```python
horse_power_median = df['horsepower'].median()
most_frequent_value = df['horsepower'].mode(dropna= True)[0]
df['horsepower_new'] = df['horsepower'].fillna(most_frequent_value)
display(horse_power_median)
display(most_frequent_value)
display(df['horsepower'].describe())
display(df['horsepower_new'].describe())
```

```python
car_asia = df[df['origin'] == 'Asia']
car_asia = car_asia[['vehicle_weight','model_year']]
car_asia_top_7 = car_asia.head(7).to_numpy()

X = car_asia_top_7
XTX = X.T @ X
XTX_invert = np.linalg.inv(XTX)
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
w = XTX_invert @ X.T @ y
display(np.sum(w))
```
